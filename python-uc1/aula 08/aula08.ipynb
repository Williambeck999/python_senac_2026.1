 # 2️⃣ Crie um programa Python que use a função main() para calcular e imprimir a soma de dois números fornecidos pelo usuário.

```python
def main():
    num_1 = float(input("digite o primeiro numero : "))
    num_2 = float(input("digite o segundo numero : "))
    result = num_1 + num_2
    print(f"a soma é :,{result}")
    
if __name__ == "__main__":
  main()
```
