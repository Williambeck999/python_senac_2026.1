 # 📚 Exercício 6️⃣: Encontrar o maior e o menor valor em um vetor

```python
vetor = [2,4,6,8,10]
vetor_maior = max(vetor)
vetor_min = min(vetor)
print ("maior valor :", vetor_maior)
print("menor valor : ", vetor_min)
```

 # 📚 Exercício 7️⃣: Inverter a ordem dos elementos de um vetor sem usar reverse()

```python
vetor =[10,20,30,40,50]
vetor_invert = vetor[::-1]
print("valor inverto :", vetor_invert)
```

 # 📚 Exercício 8️⃣: Multiplicar cada elemento do vetor por 2 e armazenar em um novo vetor

```python
vetor = [1,2,3,4,5]
multiplicador = 2
```

```python
vetor_mult =[]
for elemento in vetor :
    vetor_mult.append( elemento * multiplicador )
```

```python
print (vetor_mult)
```
